```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.naive_bayes import GaussianNB
import pickle

# --- 1. Define the Top 10 Features ---
# These features were selected based on the F-score analysis as the most significant.
TOP_10_FEATURES = [
    'social_energy', 'alone_time_preference', 'talkativeness', 'group_comfort', 
    'party_liking', 'leadership', 'risk_taking', 'public_speaking_comfort', 
    'excitement_seeking', 'reading_habit'
]

# --- 2. Load and Prepare the Data (Full 29-Feature Structure) ---
print("--- Starting Final Model Retraining (10 Features) ---")
df = pd.read_csv("personality_dataset_350_each.csv")
X = df.drop('personality_type', axis=1) # The original 29-column feature DataFrame
y = df['personality_type']

# Encode the target variable for training (re-fitting the encoder just to be safe)
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# --- 3. Fit and Save the FULL 29-Column Scaler ---
# CRITICAL: The scaler must be fitted on ALL 29 columns to work with the Streamlit app logic.
scaler = StandardScaler()
X_scaled_full = scaler.fit_transform(X)

with open('scalar.pkl', 'wb') as f:
    pickle.dump(scaler, f)
print("✅ Updated Scalar saved as 'scalar.pkl' (Fitted on ALL 29 features)")


# --- 4. Select the 10 Features from the Scaled Array ---
# Get the numerical index/position of the 10 chosen features
# We use the DataFrame X (which has the column names) to get the index.
feature_indices = [X.columns.get_loc(col) for col in TOP_10_FEATURES]

# Slice the full 29-column scaled array (X_scaled_full) to keep only the 10 columns
X_scaled_10 = X_scaled_full[:, feature_indices]

# --- 5. Retrain and Save the Naive Bayes Model (on 10 Features) ---
Naive_Bayes_10 = GaussianNB()

# Train the model only on the 10 scaled features!
Naive_Bayes_10.fit(X_scaled_10, y_encoded)

# Save the new 10-feature model (overwriting the previous model file)
with open('best_model.pkl', 'wb') as f:
    pickle.dump(Naive_Bayes_10, f)
print("✅ New Naive Bayes model saved as 'best_model.pkl' (Trained on ONLY 10 features)")

print("--- Training Script Complete. Ready for Streamlit Deployment. ---")

```
